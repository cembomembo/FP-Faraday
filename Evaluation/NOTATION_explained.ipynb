## FAR notation overview

### Magnetic field
- $B$: generic magnetic flux density
- $B(z)$: local axial field profile along the solenoid
- $B_{\mathrm{center}}$: field measured at the solenoid center
- $\bar B_{\mathrm{eff}}$: length-averaged **effective / RMS** field acting on the sample
- $\bar B_{\mathrm{Ampl}}$: length-averaged **amplitude** of the sinusoidal field

Relation:
$$
B(z)\;\rightarrow\; B_{\mathrm{center}}\;\rightarrow\; \bar B_{\mathrm{eff}}\;\rightarrow\; \bar B_{\mathrm{Ampl}}
$$

With
$$
\bar B_{\mathrm{Ampl}}=\sqrt{2}\,\bar B_{\mathrm{eff}}
$$
and in our calibration
$$
\bar B_{\mathrm{eff}} \approx 0.80\,B_{\mathrm{center}}.
$$

### Faraday effect
- $\theta$: Faraday rotation angle
- $\theta_B$: amplitude of the modulated Faraday rotation
- $V(\lambda)$: Verdet constant
- $L$: sample length / optical path length

### Refractive index / dispersion
- $n$: refractive index
- $n(\lambda)$: wavelength-dependent refractive index
- $n_l, n_r$: refractive indices for left-/right-circularly polarized light
- $\Delta n = n_l - n_r$: circular birefringence
- $\lambda_R$: effective resonance wavelength of the one-oscillator model
- $\Delta \lambda$: small Zeeman-induced shift of the resonance

### Angles / phases
- $\varphi$: analyzer angle
- $\phi$: wave phase / phase difference, depending on context

### Detector voltages
- $U$: generic detector voltage
- $U_{\mathrm{DC}}$: DC detector voltage
- $U_{\mathrm{AC}}$: AC amplitude of detector signal
- $U_{\mathrm{SS}}$: peak-to-peak detector voltage
- $U_{\parallel}$: voltage for parallel polarizers
- $U_{\perp}$: voltage for crossed polarizers
- $U_{\mathrm{bg}}$: background voltage

### Polarization / contrast
- $K$: contrast factor
$$
K=\frac{U_{\parallel}-U_{\perp}}{U_{\parallel}+U_{\perp}}
$$
- $G$: degree of polarization
$$
G=\frac{U_{\parallel}}{U_{\perp}}
$$

### Fit parameters
- $m, b$: slope and intercept of a linear fit
- $A, B$: parameters of the linearized dispersion model
- $N$: density of optically active electrons  
  (sometimes also number of data points, so context matters)

### Important consistency rule
Use:
- plain $B$ only for generic theory or $B(z)$
- $B_{\mathrm{center}}$ for raw center-field measurements
- $\bar B_{\mathrm{eff}}$ for averaged RMS field
- $\bar B_{\mathrm{Ampl}}$ for averaged amplitude

Avoid bare $\bar B$ whenever possible.
